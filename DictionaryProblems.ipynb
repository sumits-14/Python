## **Basic**

```python
# 1.Write a Python program to create a dictionary and display all key-value pairs.
d1 = {"name" : "Sumit", "age" : 26}
print(d1)
print(d1["name"])
print(d1['age'])
```

```python
# 2.Write a Python program to add a new key-value pair to a dictionary.
d2 = {"Name" : "Amit"}
d2.update({"City" : "Kolhapur"})
print(d2)
```

```python
# 3.Write a Python program to update the value of an existing key in a dictionary.
d3 = {'Name': 'Amit', 'City': 'Kolhapur', "age" : 25}
print(d3)
d3.update({
    "age" : 26
})
print(d3)
```

```python
# 4.Write a Python program to delete a key from a dictionary.
d4 = {"a": 1, "b": 2, "c": 3}
d4.pop("b")
print(d4)
```

```python
# 5.Write a Python program to check whether a key exists in a dictionary.
d5 = {'name': 'Sumit', 'age': 26}
key = "age"
if key in d5:
  print("Key exists =", True)
else:
  print("Key doesn't exits in d5")
```

```python
# 6.Write a Python program to print all keys, all values, and all items of a dictionary.
d6 = {"a": 10, "b": 20}
print("Keys = ",d6.keys())
print("Values = ", d6.values())
print("Items =", d6.items())
```

```python
# 7.Write a Python program to find the sum of all numeric values in a dictionary.
d7 = {'a' : 10, 'b' : 20, 'c' : 30}
sum(list(d7.values()))
```

```python
# 8.Write a Python program to merge two dictionaries.
d8 = {"a": 1, "b": 2}
d9 = {"c": 3, "d": 4}
print({**d8, **d9})
```

```python
# 9.Write a Python program to count the frequency of characters in a string using a dictionary.
st = 'aabccc'
d10={}
for i in st:
  if i in d10:
    d10[i] += 1
  else:
    d10[i] = 1
print(d10)
```

```python
# 10.Write a Python program to create a dictionary from two lists: one of keys and one of values.
k = ['name', 'age', 'city']
v = ['Amit', 21, 'Kolhapur']
print(dict(zip(k, v)))
```

```python
# 11.Write a Python program to sort a dictionary by keys.
d11 = {"c": 3, "a": 1, "b": 2}
dict(sorted(d11.items()))
```

```python
dict(sorted(d11.items(), reverse=True))
```

```python
# 12.Write a Python program to sort a dictionary by values
d12 = {"a": 30, "b": 10, "c": 20}
val = list(d12.values())
val.sort()
res = {}
for v in val:
  for k in d12:
    if d12[k] == v:
      res[k] = v
print(res)
```

```python
dict(sorted(d12.items(), key = lambda item : item[1] ))
```

```python
# 13.Write a Python program to find the key with the maximum value in a dictionary
d13 = {"A": 75, "B": 92, "C": 88}
m = max(list(d13.values()))
for k in d13:
  if d13[k] == m:
    print('Key with max value is :', k)
```

```python
max(d13, key=lambda key:d13[key])
```

```python
# 14.Write a Python program to convert a dictionary into a list of tuples.
d14 = {'a': 1, 'b': 2, 'c': 3}
print(list(d14.items()))
```

```python
# 15.Write a Python program to create a nested dictionary for student details.
# Student 101: Name = Amit, Marks = 85
# Student 102: Name = Neha, Marks = 90
student = {
    101 : {
        "name" : "AMit",
        "marks" : 85
    },
    102 : {
        "name" : "Neha",
        "marks" : 80
    }
}
print(student[101]['name'])
```

```python
# 16.Write a Python program to count the frequency of words in a sentence using a dictionary.
str1 = 'python is easy and python is powerful'
l = str1.split()
fr={}
for i in l:
  if i in fr:
    fr[i] += 1
  else:
    fr[i] = 1
print(fr)
```
