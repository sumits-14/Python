```python
# 1. Write a Python higher-order function that accepts another function and two numbers, then applies that function to the numbers.
a = 10
b = 20

def calculate(func, a, b):
  return func(a, b)

def add(a, b):
  return a+b

print(calculate(add, a, b))
```

```python
# 9. Write a Python function that returns another function for multiplying a value by a fixed number.
m = 5
v = 8
def result(funs, m, v):
  return funs(m, v)

def multiplication(m, v):
  return m * v

print(result(multiplication, m , v))
```

```python
# 10. Write a Python function that returns a greeting function based on the selected language.
language = 'Marathi'
name = 'Asha'
def get_greeting(language):
  def greeting():
    if language == 'English':
      return 'Hello,'
    elif language == 'Hindi':
      return 'Namaste,'
    elif language == 'Marathi':
      return 'Namaskar,'
    else:
      return 'Language not fonud!'

  return greeting

greet = get_greeting(language)

print(greet()+" "+name+"!")
```

```python
# 11. Write a Python higher-order function that accepts a transformation function and applies it to every item in a list.
data = [1, 2, 3]
trans = 'cube'
def transform(ops, d):
  def result():
    if ops == 'square':
      return list(map(lambda x : x**2, d))
    elif ops == 'cube':
      return list(map(lambda x : x**3, d))
    else:
      return 'Check data once!'

  return result
final = transform(trans, data)
print(final())
```

```python
# 12. Write a Python function that returns add, subtract, multiply, or divide function according to an operation name.
ops = 'subtract'
a = 30
b = 12
def mathOps(ops, a, b):
  def result():
    if ops == 'add':
      return a + b
    elif ops == 'subtract':
      return a - b
    elif ops == 'multiply':
      return a * b
    else:
      return a / b
  return result
final = mathOps(ops, a, b)
print(final())
```

```python
# 13. Write a Python program to filter names having at least five characters by passing a predicate function.
names = ['Asha', 'Rohan', 'Neha', 'Sameer']

def fiveOrMore(name):
  return len(name) >= 5

def filterNames(names, fiveOrMore):
  result = []

  for name in names:
    if fiveOrMore(name):
      result.append(name)

  return result

result = filterNames(names, fiveOrMore)
print(result)
```

```python
# 16. Write a Python higher-order function that calls a supplied function a given number of times and collects the returned values
word = 'Hello'
repeat = 3
def multipleCalls(func, repeat):
  result = []

  for i in range(repeat):
    r = func()
    result.append(r)

  return result

def greet():
  return word

final = multipleCalls(greet, repeat)
print(final)
```
