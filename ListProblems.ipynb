```python
# 8.Write a Python program to find the second largest number in a list.
li09 = [10, 40, 20, 30, 50]
l = 0
sl = 0
for i in li09:
  if i > l:
    sl = l
    l = i
  if sl < i and i != l:
    sl = i
print(sl)
```

```python
# 14.Write a Python program to find common elements between two lists.
li15 = [1, 2, 3, 4]
li16 = [3, 4, 5, 6]
common = []
for i in li15:
  if i in li16:
    common.append(i)
print(common)
```
