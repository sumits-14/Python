### **Basic**

```python
# 1.Write a Python program to print numbers from 1 to 10 using range().
for i in range(1, 11):
  print(i)
```

```python
# 2.Write a Python program to print all even numbers from 1 to 20 using range().
for i in range(1, 21):
  if i%2==0:
    print(i)
```

```python
# 3.Write a Python program to print all odd numbers from 1 to 20 using range().
for i  in range(0,21):
  if i % 2 != 0:
    print(i)
```

```python
# 4.Write a Python program to print numbers from 10 to 1 in reverse order using range().
for i in range(10, 0, -1):
  print(i)
```

```python
# 5.Write a Python program to print the multiplication table of a number using range().
for i in range(1, 11):
  print('5 x',i, '=', i*5)
```

```python
# 6.Write a Python program to find the sum of numbers from 1 to n using range()
sum = 0
n = 5
for i in range(1, n+1):
  sum+=i
print(sum)
```

```python
# 7.Write a Python program to find the factorial of a number using range()
n = 5
f = 1
for i in range(1, n+1):
  f*=i
print(f)
```

```python
# 8.Write a Python program to print squares of numbers from 1 to 10 using range().
for i in range(1, 11):
  print(i**2)
```

```python
# 9.Write a Python program to print cubes of numbers from 1 to 10 using range().
for i in range(1, 11):
  print(i**3)
```

```python
# 10.Write a Python program to count how many numbers between 1 and 100 are divisible by 7 using range().
c = 0
for i in range(1, 101):
  if i%7==0:
    c+=1
print(c)
```

```python
# 11.Write a Python program to print all numbers between 1 and 50 except multiples of 5 using range()
for i in range(1, 51):
  if i % 5 != 0:
    print(i)
```

```python
# 12.Write a Python program to print the Fibonacci series up to n terms using range().
n = 7
o=0
s=1
for i in range(n):
  print(o, end=" ")
  o, s = s, o + s
```

```python
# 13.Write a Python program to print the ASCII values of uppercase letters using range().
for i in range(65, 91):
  print(chr(i), '=', i)
```

```python
# 14.Write a Python program to generate a list of numbers from 100 to 0 with a step of -10 using range().
print(list(range(100, -1, -10)))
```

```python
# 15.Write a Python program to print all leap years between two given years using range().
s = 2000
e = 2020
for i in range(s, e+1):
  if i%4==0:
    print(i)
```

```python
# 16.Write a Python program to display a star pattern using nested loops and range().
r = 4
st='*'
for i in range(1, r+1):
  print(st)
  st+="*"
```
