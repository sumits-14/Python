```python
# Write a Python program to check whether two strings are anagrams.
a = 'listen'
b = 'silent'
count = 0
for i in a:
  if b.count(i):
    count+=1

if count == len(b):
  print('anagram')
else:
  print('not an anagram')
```

```python
# 16.Write a Python program to count uppercase letters, lowercase letters, digits, and special characters in a string.
str09 = "LihaTech@2026"
upper = 0
lower = 0
digits = 0
Special = 0
for i in str09:
  if i.isdigit():
    digits += 1
  elif i.isupper():
    upper += 1
  elif i.islower():
    lower += 1
  else:
    Special += 1
print("UpperCase =", upper, "Lowercase =", lower, "Digits =", digits, "Special =", Special)
```

```python
# 8. Write a Python program to compress consecutive repeated characters using character counts.
st17 = 'aaabbccccdd'
fr = {}
for i in st17:
  if i in fr:
    fr[i] += 1
  else:
    fr[i] = 1
a = list(fr.keys())
b = list(fr.values())
s = ''
for i in range(len(a)):
  s+=a[i]
  s+=str(b[i])
print(s)
```

```python
comp = ''
count = 1
for i in range(1, len(st17)):
  if st17[i] == st17[i - 1]:
    count+=1
  else:
    comp += st17[i-1] + str(count)
    count = 1
comp += st17[-1] + str(count)
print(comp)
```
