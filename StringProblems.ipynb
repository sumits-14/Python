```python
# Write a Python program to check whether two strings are anagrams.
a = 'listen'
b = 'silent'
count = 0
for i in a:
  if b.count(i):
    count+=1

if count == len(b):
  print('anagram')
else:
  print('not an anagram')
```

```python
# 10.Write a Python program to find the frequency of each character in a string
t = "aabcc"
fr = {}
for c in t:
  if c in fr:
    fr[c] += 1
  else:
    fr[c] = 1
print(fr)
```

```python
# 15.Write a Python program to remove duplicate characters from a string while preserving order
str08 = "programming"
es = ""
ess = set()
for c in str08:
  if c not in ess:
    es += c
    ess.add(c)
print(es)
```

```python
# 16.Write a Python program to count uppercase letters, lowercase letters, digits, and special characters in a string.
str09 = "LihaTech@2026"
upper = 0
lower = 0
digits = 0
Special = 0
for i in str09:
  if i.isdigit():
    digits += 1
  elif i.isupper():
    upper += 1
  elif i.islower():
    lower += 1
  else:
    Special += 1
print("UpperCase =", upper, "Lowercase =", lower, "Digits =", digits, "Special =", Special)
```
