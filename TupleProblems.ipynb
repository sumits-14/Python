## **Tuple Practice Problems**

### **Basic**

```python
# 1.Write a Python program to create a tuple and print its elements
t01 = (10, 20, 30 ,40)
print(t01)
```

```python
# 2.Write a Python program to find the length of a tuple.
t02 = (1, 2, 3, 4, 5)
print(len(t02))
```

```python
# 3.Write a Python program to count the occurrences of an element in a tuple.
t03 = (1, 2, 2, 3, 2, 4)
e = 2
count = 0
for i in t03:
  if i==e:
    count+=1
print(count)
```

```python
# 4.Write a Python program to find the index of an element in a tuple.
t04 = (10, 20, 30, 40)
e1 = 30
print("Index =",t04.index(e1))
```

```python
# 5.Write a Python program to convert a tuple into a list and a list into a tuple.
t05 = (1, 2, 3)
l01 = [4, 5, 6]
t = tuple(l01)
l = list(t05)
print("List to tuple",t)
print("Tuple to List", l)
```

```python
# 6.Write a Python program to find the maximum and minimum values in a tuple
t06 = (8, 3, 15, 1, 9)
print(max(t06), min(t06), sum(t06))
```

```python
# 7.Write a Python program to slice a tuple and print specific elements
t07 = (10, 20, 30, 40, 50)
# s = [1:4]
print(t07[1:4])
```

```python
# 8.Write a Python program to concatenate two tuples
t08 = (1, 2, 3)
t09 = (4, 5)
print(t08 + t09)
```

```python
# 9.Write a Python program to check whether an element exists in a tuple
t10 = (10, 20, 30)
e = 20
if e in t10:
  print(e, "Exists in the tuple")
```

```python
# 10.Write a Python program to create a tuple with repeated elements
e1 = 5
r = 4
# t11 = tuple(e1)
print((e1,)*r)
```

```python
# 11.Write a Python program to unpack a tuple into separate variables
t11 = (101, "Amit", 85.5)
Roll, Name, marks = t11
print(Roll, Name, marks)
```

```python
# 12.Write a Python program to sort tuple elements and display the result
t12 = (5, 2, 8, 1, 4)
print(tuple(sorted(t12)))
```

```python
# 13.Write a Python program to remove duplicate values from a tuple.
t13 = (1, 2, 2, 3, 1, 4)
l1 = list()
for i in t13:
  if i not in l1:
    l1.append(i)
print(tuple(l1))
```

```python
print(tuple(set(t13)))
```

```python
r = dict.fromkeys(t13)
print(tuple(r))
```

```python
# 14.Write a Python program to create a nested tuple and access its inner elements.
t14 = (1, 2, (10, 20, 30))
id = 0
for i in t14:
  if type(i) == tuple:
    id = t14.index(i)
    print(i)
print(t14[2][1])
```

```python
# 15.Write a Python program to swap two tuples.
t15 = (1, 2)
t16 = (3, 4)
t15, t16 = t16, t15
print(t15, t16)
```

```python
# 16.Write a Python program to find the sum of numeric elements in a tuple.
t17 = (10, 20, 30, 40)
print(sum(t17))
```
