```python
# 1.Write a Python function that accepts a name and returns a greeting message
def f1(name):
  return "Hello, "+name+"!"
f1("Asha")
```

```python
# 2.Write a Python function that accepts two numbers and returns their sum.
a = 12
b = 8
def f2(a, b):
  return a+b
print(f2(a, b))
```

```python
# 3.Write a Python function that returns the greater of two numbers.
a = 35
b = 48
def f3(a, b):
  if a>b:
    return a
  else:
    return b
print(f3(a, b))
```

```python
# 4.Write a Python function that checks whether a number is even or odd.
num = 26
def f4(n):
  if n%2==0:
    return "Even"
  else:
    return "Odd"
print(f4(num))
```

```python
# 5.Write a Python function that returns the square of a number.
num = 7
def f5(n):
  return n**2
print(f5(num))
```

```python
# 6.Write a Python function to calculate the area of a rectangle.
l = 12
w = 5
def f6(l, w):
  return l*w
print(f6(l, w))
```

```python
# 7.Write a Python function to convert Celsius temperature to Fahrenheit.
c = 25
def f7(c):
  return (c*9/5)+32
print(f7(c))
```

```python
# 8.Write a Python function to count the number of vowels in a string.
txt = 'Programming'
def f8(t):
  count=0
  for i in t:
    if i in 'aeiou':
      count+=1
  print(count)
f8(txt)
```

```python
# 9.Write a Python function to reverse a given string.
tx = 'Python'
def f9(t):
  k = list(t)
  k.reverse()
  return "".join(k)
print(f9(tx))
```

```python
def f10(t):
  return t[::-1]
print(f10(tx))
```

```python
# 10.Write a Python function to calculate the factorial of a number.
num = 5
def f11(n):
  f = 1
  while n>0:
    f*=n
    n-=1
  return f
print(f11(num))
```

```python
# 11.Write a Python function to return the sum of all numbers in a list.
l = [10, 20, 30, 40]
def f12(l):
  return sum(l)
print(f12(l))
```

```python
# 12.Write a Python function to calculate the average of numbers in a list.
n = [10, 20, 30, 40]
def f13(n):
  return sum(n)/len(n)
print(f13(n))
```

```python
# 13.Write a Python function to check whether a string is a palindrome.
t = 'madam'
def f14(t):
  return t==t[::-1]
print(f14(t))
```

```python
# 14.Write a Python function to calculate simple interest.
p = 10000
r = 8
t = 2
def f15(p, r, t):
  return (p*t*r)/100
print(f15(p,r,t))
```

```python
# 15.Write a Python function with a default greeting message that can also accept a custom message.
n = 'Ravi'
cm = input('Type a message:')
def f16(n,cm):
  return cm+' '+n+'!'
print(f16(n,cm))
```
